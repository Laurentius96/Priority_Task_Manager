# 📋 Priority Task Manager

An interactive command-line task manager with priority levels and JSON persistence.

Run the cell below to start the menu. Tasks are loaded from `tasks.json` at startup and saved with option **4**.

```python
import json
import os

TASKS_FILE = "tasks.json"
tasks = []

# Layout helpers
def clear_screen():
    os.system('cls' if os.name == 'nt' else 'clear')

def separator():
    print("=" * 60)

def header():
    clear_screen()
    separator()
    print("📋 TASK MANAGER".center(60))
    separator()

# Persistence
def save_tasks():
    with open(TASKS_FILE, "w") as file:
        json.dump(tasks, file, indent=2)
    print("\n✅ Tasks saved successfully!")
    input("\nPress ENTER to continue...")

def load_tasks():
    try:
        with open(TASKS_FILE, "r") as file:
            tasks = json.load(file)
            print("✅ Tasks loaded successfully!")
            return tasks
    except:
        print("⚠️  No task file found! Starting a new list.")
        return []

# Task operations
def add_task(name, priority):
    task = {
        "name": name,
        "priority": priority
    }

    tasks.append(task)
    print(f"\n✅ Task added successfully!")
    print(f"📝 {task['name']} | 🔥 Priority: {task['priority']}")
    input("\nPress ENTER to continue...")

def list_tasks():
    header()
    print("📋 TASK LIST")
    separator()

    if not tasks:
        print("📭 No tasks found!")
    else:
        for i, task in enumerate(tasks, 1):
            priority_icon = "🔴" if task['priority'].lower() in ['high', 'urgent'] else "🟡" if task['priority'].lower() == 'medium' else "🟢"
            print(f"{i:2d}. {priority_icon} {task['name']}")
            print(f"     └─ Priority: {task['priority']}")
            print()

    separator()
    input("Press ENTER to return to the menu...")

def remove_task(name):
    for task in tasks:
        if task["name"].lower() == name.lower():
            tasks.remove(task)
            print(f"\n✅ Task '{name}' removed successfully!")
            input("\nPress ENTER to continue...")
            return

    print(f"\n❌ Task '{name}' not found!")
    input("\nPress ENTER to continue...")

def show_menu():
    header()
    print("🎯 MENU")
    separator()
    print("1️⃣  Add task")
    print("2️⃣  List tasks")
    print("3️⃣  Remove task")
    print("4️⃣  Save tasks")
    print("5️⃣  Exit")
    separator()

# Startup
print("🚀 Starting Task Manager...")
tasks = load_tasks()
input("\nPress ENTER to continue...")

while True:
    show_menu()

    try:
        option = int(input("🔹 Choose an option (1-5): "))
    except ValueError:
        print("\n❌ Please enter numbers only!")
        input("Press ENTER to continue...")
        continue

    if option == 1:
        header()
        print("➕ ADD NEW TASK")
        separator()
        name = input("📝 Task name: ").strip()

        if not name:
            print("\n❌ Task name cannot be empty!")
            input("Press ENTER to continue...")
            continue

        print("\n🔥 Priority options:")
        print("   • High/Urgent")
        print("   • Medium")
        print("   • Low")
        priority = input("\n🎯 Task priority: ").strip()

        if not priority:
            priority = "Medium"

        add_task(name, priority)

    elif option == 2:
        list_tasks()

    elif option == 3:
        header()
        print("🗑️  REMOVE TASK")
        separator()

        if not tasks:
            print("📭 No tasks to remove!")
            input("Press ENTER to continue...")
            continue

        print("📋 Available tasks:")
        for i, task in enumerate(tasks, 1):
            print(f"   {i}. {task['name']}")
        print()

        name = input("🔹 Name of the task to remove: ").strip()
        if name:
            remove_task(name)
        else:
            print("\n❌ Name cannot be empty!")
            input("Press ENTER to continue...")

    elif option == 4:
        header()
        print("💾 SAVING TASKS...")
        separator()
        save_tasks()

    elif option == 5:
        header()
        print("👋 CLOSING PROGRAM")
        separator()
        print("Thank you for using the Task Manager!")
        print("See you soon! 🚀")
        break

    else:
        print(f"\n❌ Invalid option '{option}'! Choose between 1 and 5.")
        input("Press ENTER to continue...")
```
